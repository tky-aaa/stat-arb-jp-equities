```python
from pathlib import Path

from statarb.data.instruments.japan_equity import JapanEquity
from statarb.data.loaders.yfinance_loader import YahooFinanceSource
from statarb.data.transform.transform import preprocess_prices

# ======================================================
# Universe
# ======================================================

stocks = [
    JapanEquity("5232"),
    JapanEquity("5233"),
    JapanEquity("5269"),
]


# ======================================================
# Download
# ======================================================

source = YahooFinanceSource()

prices = source.get_prices(
    instruments=stocks,
    start="2024-01-01",
    end="2026-01-01",
)

# ======================================================
# Preprocess
# ======================================================

log_prices = preprocess_prices(
    prices,
)

print(log_prices.shape)

display(log_prices.head())

# ======================================================
# Save
# ======================================================

save_dir = Path(
    "tests/integration/data"
)

save_dir.mkdir(
    parents=True,
    exist_ok=True,
)

log_prices.to_pickle(
    save_dir / "cement_prices.pkl"
)

print(
    f"Saved to: {save_dir / 'cement_prices.pkl'}"
)
```
